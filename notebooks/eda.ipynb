```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv('student_habits_performance.csv')
df.info()

df_original = df.copy()
```

```python
df.head()
```

```python
df.isnull().sum()
```

#### Here the parental_education_level has 91 null values, so it must be filled

```python
df['parental_education_level']
```

```python
df['parental_education_level'].unique()
```

```python
df['parental_education_level'].mode()
```

```python
#We will fill the data using mode imputation technique

mode = df['parental_education_level'].mode()[0]  # [0]-gets the scalar value
df.fillna({'parental_education_level': mode}, inplace=True)
```

```python
df['parental_education_level'].isnull().sum()
```

```python
df.isnull().sum()
```

```python
df = df.drop(['student_id'],axis=1)
```

```python
df.head()
```

```python
df['gender'].unique()
```

```python
df['gender'].value_counts()
```

```python
df = pd.get_dummies(df, columns=['gender'], drop_first=True)
```

```python
df.columns.tolist()
```

```python
df.head()
```

```python
df = pd.get_dummies(df, columns=['extracurricular_participation'], drop_first=True)
```

```python
df.head()
```

```python
df['gender_Male'] = df['gender_Male'].astype(int)
```

```python
df.head()
```

```python
df['gender_Other'] = df['gender_Other'].astype(int)
df['extracurricular_participation_Yes'] = df['extracurricular_participation_Yes'].astype(int)
```

```python
df.head()
```

```python
df['part_time_job'].unique()
```

```python
df['part_time_job'] = df['part_time_job'].map({'Yes':1,'No':0})
```

```python
df.head()
```

```python
df['diet_quality'].unique()
```

```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(categories=[['Poor', 'Fair', 'Good']])
df['diet_quality'] = encoder.fit_transform(df[['diet_quality']])
```

```python
df.head()
```

```python
print(df['internet_quality'].unique())
print(df['parental_education_level'].unique())
```

```python
internet_map = {'Poor': 0, 'Average': 1, 'Good': 2}
df['internet_quality'] = df['internet_quality'].map(internet_map)

edu_map = {
    'High School': 0,
    "Bachelor": 1,
    "Master": 2,
}
df['parental_education_level'] = df['parental_education_level'].map(edu_map)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.to_csv('../data/processed_data.csv', index=False)
```
